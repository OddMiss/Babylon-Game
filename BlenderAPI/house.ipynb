```python
import bpy

# Clear existing objects
bpy.ops.object.select_all(action='SELECT')
bpy.ops.object.delete(use_global=False)

# Create the base of the house
bpy.ops.mesh.primitive_cube_add(size=4, location=(0, 0, 2))
base = bpy.context.object
base.name = "House_Base"

# Create the roof
bpy.ops.mesh.primitive_cone_add(vertices=4, radius1=3.5, depth=2, location=(0, 0, 5))
roof = bpy.context.object
roof.name = "House_Roof"
roof.rotation_euler[2] = 0.785398  # Rotate 45 degrees for pyramid shape

# Create a door
bpy.ops.mesh.primitive_cube_add(size=1, location=(0, -2.01, 0.5))
door = bpy.context.object
door.name = "House_Door"
door.scale[0] = 0.5
door.scale[2] = 1

# Create a window
bpy.ops.mesh.primitive_cube_add(size=0.5, location=(1.5, -2.01, 2))
window = bpy.context.object
window.name = "House_Window"

# Optional: Set materials/colors
mat_base = bpy.data.materials.new(name="BaseMaterial")
mat_base.diffuse_color = (0.8, 0.6, 0.4, 1)
base.data.materials.append(mat_base)

mat_roof = bpy.data.materials.new(name="RoofMaterial")
mat_roof.diffuse_color = (0.5, 0.1, 0.1, 1)
roof.data.materials.append(mat_roof)
```
